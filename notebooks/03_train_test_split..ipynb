```python
import pandas as pd
from sklearn.model_selection import train_test_split

# 1. Load the ML-ready dataset
X = pd.read_csv("../data/final/X_raw.csv")
y = pd.read_csv("../data/final/y_raw.csv")

# 2. Define Splitting Strategy (80% Train, 20% Test)
# Use a random seed for experiment reproducibility
X_train, X_test, y_train, y_test = train_test_split(
    X, y, 
    test_size=0.20, 
    random_state=42,
    stratify=y  # Maintains representative distribution
)

print(f"✅ Training samples: {len(X_train)}")
print(f"✅ Testing samples: {len(X_test)}")
```

```python
import json
import os

# Create the metadata
metadata = {
    "dataset_version": "1.0.0",
    "split_ratio": {"train": 0.8, "test": 0.2},
    "random_seed": 42,
    "target_variables": y.columns.tolist()
}

# Ensure the directory exists
os.makedirs("../ml/metadata", exist_ok=True)

# Store the artifact
with open("../ml/metadata/split_metadata.json", "w") as f:
    json.dump(metadata, f, indent=4)

print("✅ Metadata saved to ml/metadata/split_metadata.json")
```
