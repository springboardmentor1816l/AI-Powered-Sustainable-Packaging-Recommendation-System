```python
numeric_features = [
    "product_weight",
    "fragility_score",
    "moisture_sensitivity",
    "thermal_sensitivity",
    "cost_per_kg",
    "co2_factor",
    "durability_score",
    "co2_impact_index",
    "cost_efficiency_index"
]

categorical_features = [
    "product_category",
    "material_type",
    "packaging_type",
    "recyclability_category",
    "supplier_region"
]

binary_features = [
    "hazardous_material"
]
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder

# 1. Pipeline for Numeric Data: Impute missing with median -> Scale
numeric_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])

# 2. Pipeline for Categorical Data: Impute with 'missing' -> One-Hot Encode
categorical_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='constant', fill_value='Unknown')),
    ('onehot', OneHotEncoder(handle_unknown='ignore', sparse_output=False))
])

# 3. Combine into the ColumnTransformer
preprocessor = ColumnTransformer(
    transformers=[
        ('num', numeric_transformer, numeric_features),
        ('cat', categorical_transformer, categorical_features),
        ('bin', 'passthrough', binary_features) # Already 0/1, just pass them through
    ],
    remainder='drop' # This drops IDs and columns not specified
)

print("✅ Preprocessing Pipeline Object Created")
```

```python
# Apply the transformation
X_processed = preprocessor.fit_transform(df)

# Convert to DataFrame to visualize the result
# (Getting feature names from OneHotEncoder)
cat_feature_names = preprocessor.named_transformers_['cat'].get_feature_names_out(categorical_features)
all_col_names = numeric_features + list(cat_feature_names) + binary_features

df_processed = pd.DataFrame(X_processed, columns=all_col_names)

print("✅ Data successfully transformed!")
print(f"New Shape: {df_processed.shape}")
df_processed.head()
```

```python

import pandas as pd
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
import joblib
import os

# LOAD DATA
df = pd.read_csv("data/final/X_raw.csv")

# DEFINE REAL COLUMNS (Based on your preview)
numeric_features = [
    "density", 
    "cost_per_kg", 
    "recyclability_percent", 
    "biodegradability_percent", 
    "co2_factor"
]

categorical_features = [
    "industry_use_case_enc"
]

print("✅ Data loaded and column names mapped to your actual file.")
```

```python
# Create the preprocessing logic
preprocessor = ColumnTransformer(
    transformers=[
        ('num', Pipeline(steps=[
            ('imputer', SimpleImputer(strategy='median')),
            ('scaler', StandardScaler())
        ]), numeric_features),
        
        ('cat', Pipeline(steps=[
            ('imputer', SimpleImputer(strategy='most_frequent')),
            ('onehot', OneHotEncoder(handle_unknown='ignore', sparse_output=False))
        ]), categorical_features)
    ]
)

# APPLY IT
X_processed = preprocessor.fit_transform(df)

print(f"✅ Transformation Complete!")
print(f"Original shape: {df.shape}")
print(f"New processed shape: {X_processed.shape}")
```

```python
# Create the folder if it doesn't exist
os.makedirs("models/preprocessing", exist_ok=True)

# Save the pipeline artifact
joblib.dump(preprocessor, "models/preprocessing/preprocessing_pipeline.pkl")

print("💾 Success! Saved to: models/preprocessing/preprocessing_pipeline.pkl")
```
