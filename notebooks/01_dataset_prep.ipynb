```python
import pandas as pd
import numpy as np

pd.set_option("display.max_columns", None)
```

```python
DATA_PATH = "../data/final/materials_engineered.parquet"

df = pd.read_parquet(DATA_PATH)
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df["recommended_material"] = df["material_type"]
```

```python
def cost_efficiency_category(cei):
    if cei >= 66:
        return "Low-cost"
    elif cei >= 33:
        return "Medium-cost"
    else:
        return "High-cost"

df["cost_efficiency_category"] = df["CEI"].apply(cost_efficiency_category)
```

```python
FEATURES = [
    "material_type",
    "moisture_resistance_score",
    "thermal_resistance_score",
    "Load Handling Score",
    "CO2 Emission per kg (estimated)",
    "Biodegradation Time (days)",
    "Reusability (%)",
    "Recyclability Category",
    "supplier_sustainability_compliance_pct",
    "Waste Reduction Impact (%)",
    "Sustainability Target Progress (%)",
    "Cost per Unit (USD)",
    "Annual Usage (units)",
    "Total Material Weight (tons)",
    "CII",
    "CEI",
    "MSS"
]
```

```python
X = df[FEATURES]
y = df[["recommended_material", "cost_efficiency_category"]]

X.shape, y.shape
```

```python
X.to_csv("../data/final/X_raw.csv", index=False)
y.to_csv("../data/final/y_raw.csv", index=False)
```
