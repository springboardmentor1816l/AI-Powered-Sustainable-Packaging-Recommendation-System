## 1. Load the Raw Integrated Dataset
Load the engineered materials dataset from the data pipeline.

```python
# Define paths
DATA_DIR = Path('../data')
MODEL_READY_DIR = DATA_DIR / 'model_ready'
OUTPUT_DIR = DATA_DIR / 'ml_ready'
OUTPUT_DIR.mkdir(exist_ok=True)

# Load the engineered dataset
df = pd.read_csv(MODEL_READY_DIR / 'materials_engineered.csv')

print(f"✅ Dataset loaded successfully!")
print(f"   Shape: {df.shape}")
print(f"   Rows: {df.shape[0]:,} | Columns: {df.shape[1]}")
```

```python
# Display basic information
print("\n📊 Dataset Info:")
df.info()
```

```python
# Display first few rows
print("\n📋 First 5 rows:")
df.head()
```

```python
# Check for missing values
print("\n🔍 Missing Values:")
missing_summary = df.isnull().sum()
missing_summary = missing_summary[missing_summary > 0].sort_values(ascending=False)
if len(missing_summary) > 0:
    print(missing_summary)
else:
    print("✅ No missing values found!")
```

---
## 2. Define Target Variables

### Target Variable: `recommended_material`

This represents the **primary target** for our classification model.  
The model will predict the best sustainable packaging material based on product requirements.

### Possible Target Options:
1. **`material_type`** - Direct material classification (Primary Target)
2. **`sustainability_score`** - Regression target (Optional)
3. **Cost Category** - Budget-based classification (Future Enhancement)

For this initial model, we'll use **`material_type`** as our target variable.

```python
# Define the target variable
TARGET_VARIABLE = 'material_type'

print(f"\n🎯 Target Variable: {TARGET_VARIABLE}")
print(f"\n📊 Target Distribution:")
print(df[TARGET_VARIABLE].value_counts())
print(f"\n📈 Target Distribution (%):\n{df[TARGET_VARIABLE].value_counts(normalize=True) * 100}")
```

```python
# Visualize target distribution
plt.figure(figsize=(10, 6))
df[TARGET_VARIABLE].value_counts().plot(kind='bar', color='skyblue', edgecolor='black')
plt.title('Distribution of Material Types (Target Variable)', fontsize=14, weight='bold')
plt.xlabel('Material Type', fontsize=12)
plt.ylabel('Frequency', fontsize=12)
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```

---
## 3. Define Model Features

### Feature Groups:

#### **A. Material Properties** (Physical & Environmental)
- `recyclability_percent`
- `recycled_content_percent`
- `reusability_percent`
- `biodegradation_time_days`
- `end_of_life_disposal_percent`
- `carbon_footprint_kg_co2_unit`
- `co2_emission_per_kg_estimated`
- `waste_reduction_impact_percent`
- `sustainability_target_progress_percent`

#### **B. Product Packaging Requirements** (Load & Resistance)
- `load_handling_score`
- `moisture_resistance_score`
- `thermal_resistance_score`

#### **C. Cost & Operations**
- `cost_per_unit_usd`
- `annual_usage_units`
- `total_material_weight_tons`
- `supplier_sustainability_compliance_percent`

#### **D. Engineered Indices** (From Feature Engineering)
- `co2_impact_index` (CII)
- `cost_efficiency_index` (CEI)
- `material_suitability_score` (MSS)
- `overall_sustainability_score`

#### **E. Categorical Features**
- `packaging_type`
- `suitable_product_categories`
- `recommended_packaging_use_cases`
- `supplier_region`
- `recyclability_category`

---

```python
# Define feature categories

# Numeric features
MATERIAL_PROPERTIES = [
    'recyclability_percent',
    'recycled_content_percent',
    'reusability_percent',
    'biodegradation_time_days',
    'end_of_life_disposal_percent',
    'carbon_footprint_kg_co2_unit',
    'co2_emission_per_kg_estimated',
    'waste_reduction_impact_percent',
    'sustainability_target_progress_percent'
]

PACKAGING_REQUIREMENTS = [
    'load_handling_score',
    'moisture_resistance_score',
    'thermal_resistance_score'
]

COST_OPERATIONS = [
    'cost_per_unit_usd',
    'annual_usage_units',
    'total_material_weight_tons',
    'supplier_sustainability_compliance_percent'
]

ENGINEERED_INDICES = [
    'co2_impact_index',
    'cost_efficiency_index',
    'material_suitability_score',
    'overall_sustainability_score'
]

# Categorical features
CATEGORICAL_FEATURES = [
    'packaging_type',
    'suitable_product_categories',
    'recommended_packaging_use_cases',
    'supplier_region',
    'recyclability_category'
]

# All numeric features
NUMERIC_FEATURES = (
    MATERIAL_PROPERTIES + 
    PACKAGING_REQUIREMENTS + 
    COST_OPERATIONS + 
    ENGINEERED_INDICES
)

# All features (excluding target and ID columns)
ALL_FEATURES = NUMERIC_FEATURES + CATEGORICAL_FEATURES

print("\n📋 Feature Summary:")
print(f"   Material Properties: {len(MATERIAL_PROPERTIES)} features")
print(f"   Packaging Requirements: {len(PACKAGING_REQUIREMENTS)} features")
print(f"   Cost & Operations: {len(COST_OPERATIONS)} features")
print(f"   Engineered Indices: {len(ENGINEERED_INDICES)} features")
print(f"   Categorical Features: {len(CATEGORICAL_FEATURES)} features")
print(f"\n   Total Numeric Features: {len(NUMERIC_FEATURES)}")
print(f"   Total Features: {len(ALL_FEATURES)}")
```

```python
# Display feature groups
print("\n" + "="*80)
print("📌 FINAL FEATURE LIST")
print("="*80)

print("\n🔹 Material Properties:")
for i, feat in enumerate(MATERIAL_PROPERTIES, 1):
    print(f"   {i}. {feat}")

print("\n🔹 Packaging Requirements:")
for i, feat in enumerate(PACKAGING_REQUIREMENTS, 1):
    print(f"   {i}. {feat}")

print("\n🔹 Cost & Operations:")
for i, feat in enumerate(COST_OPERATIONS, 1):
    print(f"   {i}. {feat}")

print("\n🔹 Engineered Indices:")
for i, feat in enumerate(ENGINEERED_INDICES, 1):
    print(f"   {i}. {feat}")

print("\n🔹 Categorical Features:")
for i, feat in enumerate(CATEGORICAL_FEATURES, 1):
    print(f"   {i}. {feat}")

print("\n" + "="*80)
```

---
## 4. Data Quality Checks
Verify all features exist in the dataset and check data types.

```python
# Check if all features exist in the dataset
print("\n🔍 Feature Availability Check:")
missing_features = [feat for feat in ALL_FEATURES if feat not in df.columns]

if missing_features:
    print(f"⚠️  Warning: {len(missing_features)} features not found in dataset:")
    for feat in missing_features:
        print(f"   - {feat}")
else:
    print("✅ All features are available in the dataset!")

# Check if target exists
if TARGET_VARIABLE not in df.columns:
    print(f"\n❌ Error: Target variable '{TARGET_VARIABLE}' not found!")
else:
    print(f"\n✅ Target variable '{TARGET_VARIABLE}' found!")
```

```python
# Check data types
print("\n📊 Feature Data Types:")
print("\nNumeric Features:")
for feat in NUMERIC_FEATURES:
    if feat in df.columns:
        print(f"   {feat}: {df[feat].dtype}")

print("\nCategorical Features:")
for feat in CATEGORICAL_FEATURES:
    if feat in df.columns:
        print(f"   {feat}: {df[feat].dtype} ({df[feat].nunique()} unique values)")
```

---
## 5. Prepare X (Features) and y (Target)
Separate features from the target variable.

```python
# Remove any features that don't exist in the dataset
valid_features = [feat for feat in ALL_FEATURES if feat in df.columns]

# Create X (features) and y (target)
X_raw = df[valid_features].copy()
y_raw = df[TARGET_VARIABLE].copy()

print("\n✅ Features and Target Separated:")
print(f"   X_raw shape: {X_raw.shape}")
print(f"   y_raw shape: {y_raw.shape}")
print(f"\n   Features used: {len(valid_features)}")
print(f"   Target classes: {y_raw.nunique()}")
```

```python
# Display X_raw summary
print("\n📋 X_raw (Features) - First 5 rows:")
X_raw.head()
```

```python
# Display y_raw summary
print("\n🎯 y_raw (Target) - Distribution:")
print(y_raw.value_counts())
print(f"\nTarget distribution (%):\n{y_raw.value_counts(normalize=True) * 100}")
```

---
## 6. Basic Exploratory Data Analysis

```python
# Numeric feature statistics
print("\n📊 Numeric Features - Summary Statistics:")
numeric_cols = [col for col in valid_features if col in NUMERIC_FEATURES]
X_raw[numeric_cols].describe().T
```

```python
# Categorical feature distributions
print("\n📊 Categorical Features - Value Counts:")
categorical_cols = [col for col in valid_features if col in CATEGORICAL_FEATURES]
for col in categorical_cols:
    print(f"\n{col}:")
    print(X_raw[col].value_counts())
    print(f"Unique values: {X_raw[col].nunique()}")
```

```python
# Correlation analysis for numeric features
print("\n📈 Correlation Analysis:")
plt.figure(figsize=(14, 10))
corr_matrix = X_raw[numeric_cols].corr()
sns.heatmap(corr_matrix, annot=False, cmap='coolwarm', center=0, 
            linewidths=0.5, cbar_kws={"shrink": 0.8})
plt.title('Correlation Matrix - Numeric Features', fontsize=14, weight='bold', pad=20)
plt.tight_layout()
plt.show()
```

```python
# Feature distributions
print("\n📊 Key Feature Distributions:")
fig, axes = plt.subplots(2, 2, figsize=(14, 10))
axes = axes.ravel()

# Plot top 4 engineered indices
for i, col in enumerate(ENGINEERED_INDICES[:4]):
    axes[i].hist(X_raw[col], bins=30, color='steelblue', edgecolor='black', alpha=0.7)
    axes[i].set_title(col, fontsize=12, weight='bold')
    axes[i].set_xlabel('Value')
    axes[i].set_ylabel('Frequency')
    axes[i].grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()
```

---
## 7. Save Processed Data
Save X_raw and y_raw for the next stages (encoding, scaling, feature engineering).

```python
# Save X_raw and y_raw
X_raw.to_csv(OUTPUT_DIR / 'X_raw.csv', index=False)
y_raw.to_csv(OUTPUT_DIR / 'y_raw.csv', index=False)

print("\n✅ Data Saved Successfully!")
print(f"   Location: {OUTPUT_DIR}")
print(f"   - X_raw.csv: {X_raw.shape}")
print(f"   - y_raw.csv: {y_raw.shape}")
```

```python
# Save feature metadata
feature_metadata = {
    'target_variable': TARGET_VARIABLE,
    'total_features': len(valid_features),
    'numeric_features': len([f for f in valid_features if f in NUMERIC_FEATURES]),
    'categorical_features': len([f for f in valid_features if f in CATEGORICAL_FEATURES]),
    'material_properties': MATERIAL_PROPERTIES,
    'packaging_requirements': PACKAGING_REQUIREMENTS,
    'cost_operations': COST_OPERATIONS,
    'engineered_indices': ENGINEERED_INDICES,
    'categorical_features': CATEGORICAL_FEATURES,
    'all_features': valid_features
}

import json
with open(OUTPUT_DIR / 'feature_metadata.json', 'w') as f:
    json.dump(feature_metadata, f, indent=2)

print("\n✅ Feature Metadata Saved!")
print(f"   Location: {OUTPUT_DIR / 'feature_metadata.json'}")
```
