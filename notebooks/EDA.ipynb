```python

import numpy as np
import pandas as pd
import numpy as np
from sklearn.impute import SimpleImputer


df = pd.read_csv(r"D:\ecopack_ai_project\AI-Powered-Sustainable-Packaging-Recommendation-System\data\processed\cleaned_integrated_materials.csv")


# 1. Remove rows with excessive missing values
df = df[df.isnull().mean(axis=1) < 0.3]

# Identify columns
num_cols = df.select_dtypes(include=[np.number]).columns
cat_cols = df.select_dtypes(exclude=[np.number]).columns

# 2. Numeric imputation (median)
num_imputer = SimpleImputer(strategy="median")
df[num_cols] = num_imputer.fit_transform(df[num_cols])

# 3. Categorical imputation (most frequent)
cat_imputer = SimpleImputer(strategy="most_frequent")
df[cat_cols] = cat_imputer.fit_transform(df[cat_cols])

df.isnull().sum()
```

```python
import pandas as pd
from sklearn.impute import SimpleImputer

# Load your dataset
df = pd.read_csv("EcoPackAI_dataset.csv")

# Identify numeric and categorical columns
num_cols = df.select_dtypes(include=['int64', 'float64']).columns
cat_cols = df.select_dtypes(include=['object']).columns

# Imputers (won't change anything since no missing values)
num_imputer = SimpleImputer(strategy="median")
cat_imputer = SimpleImputer(strategy="most_frequent")

# SAFE assignment using .loc
df.loc[:, num_cols] = num_imputer.fit_transform(df[num_cols])
df.loc[:, cat_cols] = cat_imputer.fit_transform(df[cat_cols])

# Save cleaned dataset
df.to_csv("cleaned_integrated_materials.csv", index=False)
```

```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder, MinMaxScaler, LabelEncoder
import joblib
import os

# ============================
# 1. LOAD CLEANED DATASET
# ============================
df = pd.read_csv("cleaned_integrated_materials.csv")
df.head()


# ============================
# 2. IDENTIFY COLUMNS
# ============================

categorical_cols = [
    "Packaging Type",
    "Material Type",
    "Recyclability Category",
    "Supplier Region",
    "Suitable Product Categories",
    "Recommended Packaging Use Cases"
]

numeric_cols = df.select_dtypes(include=["float64", "int64"]).columns.tolist()

# Remove Material ID from numeric scaling
if "Material ID" in numeric_cols:
    numeric_cols.remove("Material ID")

print("Categorical Columns:", categorical_cols)
print("Numeric Columns:", numeric_cols)


# ============================
# 3. ENCODE CATEGORICAL FEATURES
# ============================

# One-hot encode (recommended for most ML models)
ohe = OneHotEncoder(sparse_output=False, handle_unknown='ignore')

encoded_array = ohe.fit_transform(df[categorical_cols])
encoded_df = pd.DataFrame(encoded_array, columns=ohe.get_feature_names_out(categorical_cols))

# Save encoder
os.makedirs("models/encoders", exist_ok=True)
joblib.dump(ohe, "models/encoders/ohe_encoder.pkl")


# ============================
# 4. NORMALIZE NUMERIC FEATURES
# ============================

scaler = MinMaxScaler()
scaled_array = scaler.fit_transform(df[numeric_cols])
scaled_df = pd.DataFrame(scaled_array, columns=numeric_cols)

# Save scaler
os.makedirs("models/scalers", exist_ok=True)
joblib.dump(scaler, "models/scalers/numeric_scaler.pkl")


# ============================
# 5. COMBINE ENCODED + SCALED DATA
# ============================

# Keep the ID column if you need it for ML tracking
final_df = pd.concat([df["Material ID"], scaled_df, encoded_df], axis=1)

# Export final ML-ready dataset
os.makedirs("data/model_ready", exist_ok=True)
final_df.to_csv("data/model_ready/materials_final_encoded.csv", index=False)

print("Encoding + Normalization Complete!")
print("Saved: data/model_ready/materials_final_encoded.csv")
```
