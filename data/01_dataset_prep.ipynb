# EcoPackAI — Dataset Preparation for Model Development

This notebook prepares the dataset for model development by:
- selecting appropriate target variables
- identifying valid input features
- preventing data leakage
- creating feature (`X`) and target (`y`) datasets

No feature engineering or model training is performed in this notebook.


```python
import pandas as pd
from pathlib import Path
```

```python
DATA_PATH = Path("./final/materials_cleaned.parquet")

df = pd.read_parquet(DATA_PATH)

df.shape
```

```python
df.columns.tolist()
```

```python
df.head()
```

## Dataset Overview

This dataset contains material-level information related to:
- sustainability metrics
- cost and usage patterns
- performance and resistance scores
- supplier compliance
- product category and use-case applicability

The dataset is clean (no missing values) and ready for target/feature definition.


## Target Variable Identification

Based on the EcoPackAI project objective — recommending sustainable and
cost-effective packaging materials — the following composite outcomes are
identified as candidate targets:

- Final_Recommendation_Score
- CO2_Impact_Index
- Cost_Efficiency_Index
- Material_Suitability_Score

These targets represent continuous scores and are suitable for regression tasks.


```python
# Target variables are defined conceptually at this stage.
# They will be materialized after feature engineering.

PRIMARY_TARGET = "Final_Recommendation_Score"

SECONDARY_TARGETS = [
    "CO2_Impact_Index",
    "Cost_Efficiency_Index",
    "Material_Suitability_Score"
]

PRIMARY_TARGET, SECONDARY_TARGETS
```

```python
EXCLUDED_COLUMNS = [
    "Material ID",
    PRIMARY_TARGET,
    *SECONDARY_TARGETS
]

EXCLUDED_COLUMNS
```

### Why these columns are excluded
- **Material ID**: Identifier, no predictive value
- **Target variables**: Excluded to prevent target leakage


```python
FEATURE_COLUMNS = [col for col in df.columns if col not in EXCLUDED_COLUMNS]

len(FEATURE_COLUMNS)
```

```python
FEATURE_COLUMNS
```

## Feature Groups Used

The selected features include:

- Sustainability indicators (recyclability, carbon footprint, waste reduction)
- Performance scores (load handling, moisture and thermal resistance)
- Cost and usage metrics
- Supplier sustainability compliance
- Product category compatibility
- Use-case applicability

These features are available before decision-making and do not introduce leakage.


```python
X = df[FEATURE_COLUMNS]

X.shape
```

```python
# Ensure no missing values
# Validate features only (targets not yet materialized)
X.isnull().sum().sum()
```

```python
# Ensure target not present in features
PRIMARY_TARGET in X.columns
```

```python
X.head()
```

```python
from pathlib import Path

OUTPUT_DIR = Path("./model_input")
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

X.to_csv(OUTPUT_DIR / "X_raw.csv", index=False)

print("X dataset saved successfully.")
print("Target (y) will be generated after feature engineering.")
```
