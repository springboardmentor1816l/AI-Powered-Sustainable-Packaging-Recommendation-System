```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder, StandardScaler
```

```python
materials = pd.read_csv("C:\\Users\\sneha\\Desktop\\AI-Powered-Sustainable-Packaging-Recommendation-System\\data\\processed\\materials_cleaned.csv")
products = pd.read_csv("C:\\Users\\sneha\\Desktop\\AI-Powered-Sustainable-Packaging-Recommendation-System\\data\\processed\\products_cleaned.csv")
```

```python
print(materials.columns)
```

```python
materials_num = materials.select_dtypes(include=["int64", "float64"])
```

```python
materials_cat = materials.select_dtypes(include=["object"])
```

```python
from sklearn.preprocessing import LabelEncoder

for col in materials_cat.columns:
    le = LabelEncoder()
    materials_cat[col] = le.fit_transform(materials_cat[col].astype(str))
```

```python
materials_ml = pd.concat([materials_num, materials_cat], axis=1)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
materials_scaled = scaler.fit_transform(materials_ml)

materials_final = pd.DataFrame(
    materials_scaled,
    columns=materials_ml.columns
)
```

```python
materials_final.to_csv("C:\\Users\\sneha\\Desktop\\AI-Powered-Sustainable-Packaging-Recommendation-System\\data\\final\\materials_ml_ready.csv")
```

```python
import pandas as pd
import joblib
from sklearn.preprocessing import LabelEncoder, StandardScaler

materials = pd.read_csv("C:\\Users\\sneha\\Desktop\\AI-Powered-Sustainable-Packaging-Recommendation-System\\data\\processed\\materials_cleaned.csv")
```

```python
num_cols = materials.select_dtypes(include=["int64", "float64"]).columns
cat_cols = materials.select_dtypes(include=["object"]).columns
```

```python
encoders = {}

for col in cat_cols:
    le = LabelEncoder()
    materials[col] = le.fit_transform(materials[col].astype(str))
    encoders[col] = le

joblib.dump(encoders, "C:\\Users\\sneha\\Desktop\\AI-Powered-Sustainable-Packaging-Recommendation-System\\models\\encoders\\material_encoders.pkl")
```

```python
scaler = StandardScaler()
materials[num_cols] = scaler.fit_transform(materials[num_cols])

joblib.dump(scaler, "C:\\Users\\sneha\\Desktop\\AI-Powered-Sustainable-Packaging-Recommendation-System\\models\\scalers\\material_scaler.pkl")
```

```python
materials_final.to_csv(
    "C:\\Users\\sneha\\Desktop\\AI-Powered-Sustainable-Packaging-Recommendation-System\\data\\model_ready\\material_final_encoded.csv",
    index=False
)

materials_final.to_parquet(
    "C:\\Users\\sneha\\Desktop\\AI-Powered-Sustainable-Packaging-Recommendation-System\\data\\model_ready\\mterials_final_encoded.parquet",
    index=False
)
```
