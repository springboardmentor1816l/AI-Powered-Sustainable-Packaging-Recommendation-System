```python
import pandas as pd
import numpy as np
```

```python
materials = pd.read_csv("C:\\Users\\sneha\\Desktop\\AI-Powered-Sustainable-Packaging-Recommendation-System\\data\\raw\\material_dataset.csv")
products = pd.read_csv("C:\\Users\\sneha\\Desktop\\AI-Powered-Sustainable-Packaging-Recommendation-System\\data\\raw\\product_dataset.csv")
```

```python
materials.isnull().sum()
```

```python
products.isnull().sum()
```

```python
numeric_cols_materials = materials.select_dtypes(include=[np.number])

materials[numeric_cols_materials.columns] = numeric_cols_materials.fillna(
    numeric_cols_materials.mean()
)
```

```python
materials = materials.fillna("Unknown")
```

```python
numeric_cols_products = products.select_dtypes(include=[np.number])

products[numeric_cols_products.columns] = numeric_cols_products.fillna(
    numeric_cols_products.mean()
)
```

```python
products = products.fillna("Unknown")
```

```python
products["fragility_index"] = products["fragility_index"].astype(int)
```

```python
materials.drop_duplicates(inplace=True)
products.drop_duplicates(inplace=True)
```

```python
materials.info()
products.info()
```

## Data Cleaning Summary

- Loaded raw material and product datasets
- Filled missing numerical values with mean
- Filled missing categorical values with "Unknown"
- Corrected data types
- Removed duplicate records
- Saved cleaned datasets for further processing



```python
materials.to_csv("C:\\Users\\sneha\\Desktop\\AI-Powered-Sustainable-Packaging-Recommendation-System\\data\\processed\\materials_cleaned.csv", index=False)
products.to_csv("C:\\Users\\sneha\\Desktop\AI-Powered-Sustainable-Packaging-Recommendation-System\\data\\processed\\products_cleaned.csv", index=False)
```
