```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import OneHotEncoder, MinMaxScaler
import joblib
import os
```

```python
df = pd.read_csv("data/processed/cleaned_EcoPackAi_dataset.csv")
df.head()
```

```python
numeric_cols = df.select_dtypes(include=['int64','float64']).columns.tolist()
categorical_cols = df.select_dtypes(include=['object']).columns.tolist()

print("Numeric Columns:", numeric_cols)
print("Categorical Columns:", categorical_cols)
```

```python

ohe = OneHotEncoder(handle_unknown='ignore', sparse_output=False)

encoded_array = ohe.fit_transform(df[categorical_cols])


encoded_df = pd.DataFrame(
    encoded_array,
    columns=ohe.get_feature_names_out(categorical_cols)
)

encoded_df.head()
```

```python
os.makedirs("models/encoders", exist_ok=True)
joblib.dump(ohe, "C:\\Users\\sneha\\Desktop\\AI-Powered-Sustainable-Packaging-Recommendation-System\\models\\encoders\\ohe_encoder.pkl")

print("OHE encoder saved successfully!")
```

```python
scaler = MinMaxScaler()

scaled_array = scaler.fit_transform(df[numeric_cols])
scaled_df = pd.DataFrame(scaled_array, columns=numeric_cols)

scaled_df.head()
```

```python
os.makedirs("models/scalers", exist_ok=True)
joblib.dump(scaler, "C:\\Users\\sneha\\Desktop\\AI-Powered-Sustainable-Packaging-Recommendation-System\\models\\scalers\\minmax_scaler.pkl")

print("Scaler saved successfully!")
```

```python
# Drop old categorical columns
df_numeric_only = df.drop(columns=categorical_cols)

# Combine scaled numeric + encoded categorical
final_df = pd.concat([scaled_df, encoded_df], axis=1)

final_df.head()
```

```python


final_df.to_csv("C:\\Users\\sneha\\Desktop\\AI-Powered-Sustainable-Packaging-Recommendation-System\\data\\model_ready\\materials_final_encoded.csv", index=False)

print("Model-ready dataset saved successfully!")
```
