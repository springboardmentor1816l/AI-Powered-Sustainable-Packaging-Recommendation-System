```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
from pathlib import Path
import os
```

```python
from pathlib import Path

# Get project root (one level up from notebook folder)
PROJECT_ROOT = Path.cwd().parent
RAW = PROJECT_ROOT / "data" / "raw"
PROCESSED = PROJECT_ROOT / "data" / "processed"
NOTEBOOKS = PROJECT_ROOT / "notebooks"
PROCESSED.mkdir(parents=True, exist_ok=True)

DATA_PATH = RAW / "EcoPackAI_dataset.csv"
print("DATA PATH:", DATA_PATH)
print("File exists?", DATA_PATH.exists())
```

```python
df = pd.read_csv(DATA_PATH)
print("Loaded dataset. Shape:", df.shape)
print(df.head())
```

```python
# Structure overview
print("Columns:", list(df.columns))
print("\nData types:\n", df.dtypes)
display(df.head(5))
display(df.tail(5))
```

```python
# Numeric summary
display(df.describe(include=[np.number]).T)

# Categorical summary
cat_cols = df.select_dtypes(include=['object', 'category']).columns.tolist()
for c in cat_cols:
    print(f"\n--- {c} (unique: {df[c].nunique()}) ---")
    display(df[c].value_counts().head(10))
```

```python
missing = df.isna().sum().sort_values(ascending=False)
missing_pct = (missing / len(df) * 100).round(2)
mv = pd.concat([missing, missing_pct], axis=1)
mv.columns = ['missing_count', 'missing_percent']
display(mv.head(50))
mv.to_csv(PROCESSED / "missing_value_table.csv")
print("Saved missing_value_table.csv ->", PROCESSED / "missing_value_table.csv")
```

```python
dup_count = df.duplicated().sum()
print("Total duplicate rows:", dup_count)

# If duplicates exist, show sample duplicates
if dup_count > 0:
    display(df[df.duplicated(keep=False)].head(10))
```

```python
# Examples of invalid range checks - customize column names if different in your CSV
checks = []

# Replace names below with actual column names present in your dataset
# Common checks: negative costs, recyclability >100, density <=0, weight <=0
if 'cost_per_kg' in df.columns:
    bad_cost = df[df['cost_per_kg'] < 0]
    checks.append(('negative_cost_rows', len(bad_cost)))
if 'recyclability_percent' in df.columns:
    bad_recycl = df[df['recyclability_percent'] > 100]
    checks.append(('recyclability_gt_100', len(bad_recycl)))
for col in ['density','weight','volume','fragility']:
    if col in df.columns:
        bad = df[df[col] <= 0]
        checks.append((f'{col}_le_0', len(bad)))

print("Quick invalid-range checks:")
for name, cnt in checks:
    print(f" - {name}: {cnt}")
```

```python
numeric_cols = df.select_dtypes(include=[np.number]).columns.tolist()
outlier_summary = {}
for col in numeric_cols:
    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr
    outliers = df[(df[col] < lower) | (df[col] > upper)]
    outlier_summary[col] = len(outliers)

outlier_df = pd.Series(outlier_summary).sort_values(ascending=False).to_frame('outlier_count')
display(outlier_df.head(30))
outlier_df.to_csv(PROCESSED / "outlier_counts.csv")
print("Saved outlier_counts.csv ->", PROCESSED / "outlier_counts.csv")
```

```python
# Z-score based extreme values for numeric columns
z_thresh = 3
z_outliers = {}
for col in numeric_cols:
    col_nonan = df[col].dropna()
    if col_nonan.size > 0:
        z = np.abs(stats.zscore(col_nonan, nan_policy='omit'))
        z_count = (z > z_thresh).sum()
        z_outliers[col] = int(z_count)
z_df = pd.Series(z_outliers).sort_values(ascending=False).to_frame('z_outlier_count')
display(z_df.head(30))
z_df.to_csv(PROCESSED / "z_outlier_counts.csv")
```

```python
import math

# Histograms for numeric cols (save one big figure)
num_cols = len(numeric_cols)
ncols = 3
nrows = math.ceil(num_cols / ncols)  # This ensures enough rows

plt.figure(figsize=(14, 10))
df[numeric_cols].hist(bins=30, layout=(nrows, ncols), figsize=(14, 10))
plt.tight_layout()
plt.savefig(PROCESSED / "numerical_histograms.png")
plt.close()
print("Saved numerical_histograms.png")

# Boxplot (all numeric columns)
plt.figure(figsize=(14, 6))
sns.boxplot(data=df[numeric_cols], orient="h")
plt.title("Boxplots - numeric features")
plt.tight_layout()
plt.savefig(PROCESSED / "outliers_boxplot.png")
plt.close()
print("Saved outliers_boxplot.png")
```
