```python
import pandas as pd
import numpy as np
from sklearn.impute import SimpleImputer
```

```python
dfm=pd.read_csv(".\data\materials_dataset.csv")
dfm.shape
```

```python
dfm.head()
```

```python
dfm.isnull().sum()
```

```python
numeric_cols = ['Recyclability (%)', 'Recycled Content (%)', 'Reusability (%)', 'Biodegradation Time (days)', 'Carbon Footprint (kg CO2/unit)', 'CO2 Emission per kg (estimated)', 'Cost per Unit (USD)', 'Waste Reduction Impact (%)', 'Load Handling Score', 'Moisture Resistance Score', 'Thermal Resistance Score', 'Supplier Sustainability Compliance (%)', 'Annual Usage (units)', 'Total Material Weight (tons)']
categorical_cols = ['Material Type', 'Supplier Region', 'Suitable Product Categories', 'Recommended Packaging Use Cases', 'Packaging Type', 'Recyclability Category']
ID_COL = 'Material ID'
all_process_cols = categorical_cols + numeric_cols 
df_process = dfm[[ID_COL] + all_process_cols].copy()
```

```python
missing_info = dfm[numeric_cols + categorical_cols].isnull().sum()
missing_report = f"""
# Missing Value Report
---

## Summary of Missing Values

| Column | Missing Count | Strategy |
| :--- | :--- | :--- |
{'\n'.join([f"| {col} | {count} | Median Imputation (Numeric)" for col, count in missing_info.items() if count > 0 and col in numeric_cols])}
{'\n'.join([f"| {col} | {count} | Mode Imputation (Categorical)" for col, count in missing_info.items() if count > 0 and col in categorical_cols])}
## Decisions
1. **Numeric (continuous) features** (Strength, Cost, CO2, etc.): Impute using the **Median** strategy to minimize the impact of potential outliers, as per the project plan[cite: 218].
2. **Categorical features** (Material Type, Industry Use Case): Impute using the **Mode (most frequent)** strategy, although no missing values were detected in these columns in the current raw dataset.
3. **No rows were removed** as the percentage of missing data (approx. 5%) was low and spread across columns.
"""

# Save Missing Value Report (Deliverable)
with open('missing_value_report.md', 'w') as f:
    f.write(missing_report)
print(f"Generated missing value report: {'missing_value_report.md'}")
```

```python
from sklearn.impute import SimpleImputer
```

```python
num_imputer = SimpleImputer(strategy="median")
cat_imputer = SimpleImputer(strategy="most_frequent")
```

```python
df_process[numeric_cols] = num_imputer.fit_transform(df_process[numeric_cols])
```

```python
df_process[categorical_cols] = cat_imputer.fit_transform(df_process[categorical_cols])
```

```python
df_process[all_process_cols].isnull().sum()
```

```python
df_process.to_csv('cleaned_integrated_materials.csv',index=False)
```
