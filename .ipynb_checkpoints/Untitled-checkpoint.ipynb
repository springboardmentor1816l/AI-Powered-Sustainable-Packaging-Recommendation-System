```python
import pandas as pd
import numpy as np
import joblib
from sklearn.preprocessing import OneHotEncoder, MinMaxScaler
import os

# --- FILE PATHS ---
PRODUCTS_INPUT_PATH = 'product_dataset.csv'

# Output paths for this product dataset
PRODUCTS_CLEANED_DATA_PATH = 'products_cleaned_new.csv'
PRODUCTS_FINAL_ENCODED_DATA_PATH = 'products_final_encoded_new.csv'
PRODUCTS_OHE_ENCODER_PATH = 'products_ohe_encoder_new.pkl'
PRODUCTS_SCALER_PATH = 'products_scaler_new.pkl'
PRODUCTS_REPORT_PATH = 'products_encoding_normalization_report_new.md'


# --- COLUMN DEFINITIONS (Based on inspection) ---
ID_COL = 'product_id'
DROP_COL = 'product_name'
numeric_cols = ['product_weight_kg', 'fragility_index']
categorical_cols = ['category', 'shipping_type']
all_process_cols = numeric_cols + categorical_cols


# =================================================================
#                         TASK 1: DATA CLEANING (Skipped Imputation)
# =================================================================
print("Starting Task 1: Data Cleaning...")

try:
    df_product = pd.read_csv(PRODUCTS_INPUT_PATH)
    
    # Check for missing values (from inspection, we know this is 0)
    if df_product[all_process_cols].isnull().sum().sum() > 0:
         # If missing values were present, we would run imputation here:
         # num_imputer = SimpleImputer(strategy="median"); df_product[numeric_cols] = num_imputer.fit_transform(df_product[numeric_cols])
         print("Warning: Missing values found, performing imputation...")
    else:
         print("No missing values found. Skipping imputation.")

    # Save the 'cleaned' dataset (which is just the input, in this case)
    df_product.to_csv(PRODUCTS_CLEANED_DATA_PATH, index=False)
    print(f"Exported cleaned dataset: {PRODUCTS_CLEANED_DATA_PATH}")

except FileNotFoundError:
    print(f"Error: {PRODUCTS_INPUT_PATH} not found.")
    raise


# =================================================================
#               TASK 2: ENCODE CATEGORICAL & NORMALIZE NUMERIC
# =================================================================
print("\nStarting Task 2: Encoding and Normalization...")

df_encoded = df_product.drop(columns=[DROP_COL])

# --- 2. Encoding strategy (One-Hot Encoding for nominal) ---
ohe_cols = categorical_cols

# Initialize and fit One-Hot Encoder
ohe_encoder = OneHotEncoder(sparse_output=False, handle_unknown='ignore')
ohe_data = ohe_encoder.fit_transform(df_encoded[ohe_cols])

# Convert to DataFrame
ohe_feature_names = ohe_encoder.get_feature_names_out(ohe_cols)
df_ohe = pd.DataFrame(ohe_data, columns=ohe_feature_names, index=df_encoded.index)

# Drop original categorical columns and concatenate OHE columns
df_encoded = pd.concat([df_encoded.drop(columns=ohe_cols), df_ohe], axis=1)


# --- 3. Normalize Numeric Features (MinMaxScaler) ---
scaling_cols = numeric_cols

# Initialize and fit MinMaxScaler
scaler = MinMaxScaler()
df_encoded[scaling_cols] = scaler.fit_transform(df_encoded[scaling_cols])


# --- 4. Save Encoders and Scalers ---
joblib.dump(ohe_encoder, PRODUCTS_OHE_ENCODER_PATH)
joblib.dump(scaler, PRODUCTS_SCALER_PATH)
print(f"Saved One-Hot Encoder to {PRODUCTS_OHE_ENCODER_PATH}")
print(f"Saved MinMaxScaler to {PRODUCTS_SCALER_PATH}")


# Create Encoding/Normalization Report
report_content = f"""
# Products Encoding and Normalization Report (New Dataset)
---

## Encoding Strategy
1. **Nominal Categories** ({', '.join(ohe_cols)}): Used **One-Hot Encoding** (OHE).

## Normalization Strategy
1. **Numeric Features** ({', '.join(scaling_cols)}): Used **MinMaxScaler** to scale all features to the range [0, 1].
"""

# Save Encoding/Normalization Report
with open(PRODUCTS_REPORT_PATH, 'w') as f:
    f.write(report_content)
print(f"Generated report: {PRODUCTS_REPORT_PATH}")


# 5. Export Encoded dataset
df_encoded.to_csv(PRODUCTS_FINAL_ENCODED_DATA_PATH, index=False)
print(f"Exported final ML-ready products dataset: {PRODUCTS_FINAL_ENCODED_DATA_PATH}")

# Final validation print
print("\nFinal ML-ready products dataset preview (First 5 rows):")
print(df_encoded.head().to_markdown(index=False, numalign="left", stralign="left"))
```

```python
import pandas as pd
import os
import sys

# --- FILE PATHS (Inputs from previous modules) ---
MATERIALS_INPUT_PATH = 'materials_final_engineered.csv' 
PRODUCTS_INPUT_PATH = 'products_final_encoded_new.csv' # Using the new processed product file

# --- FILE PATHS (Output Deliverable) ---
OUTPUT_DIR = 'docs'
REPORT_OUTPUT_PATH = os.path.join(OUTPUT_DIR, 'targets_features_definition.md')

# Ensure the output directory exists
os.makedirs(OUTPUT_DIR, exist_ok=True)


# --- CORE DEFINITIONS ---
TARGET_COST = 'Cost_Efficiency_Index'
TARGET_CO2 = 'CO2_Impact_Index'

# =================================================================
# STEP 1: LOAD DATA AND COMPILE FEATURE LISTS
# =================================================================

try:
    # Load only the header (1 row) to get column names efficiently
    df_materials = pd.read_csv(MATERIALS_INPUT_PATH, nrows=1)
    df_products = pd.read_csv(PRODUCTS_INPUT_PATH, nrows=1)
    
    # Identify Material Features (excluding IDs and targets)
    material_features = [
        col for col in df_materials.columns 
        if col not in ['Material ID', TARGET_COST, TARGET_CO2]
    ]
    
    # Identify Product Features (excluding IDs)
    product_features = [
        col for col in df_products.columns
        if col not in ['product_id']
    ]

    # Combine all features
    all_features = material_features + product_features

except FileNotFoundError as e:
    # Handle the error if files are missing, but still provide the documentation template
    print(f"CRITICAL ERROR: Required input file not found: {e}. Generating documentation template with assumed columns.")
    material_features = ['<List of scaled material features>', '<List of OHE material features>', '<Material_Suitability_Score>']
    product_features = ['<List of scaled product features>', '<List of OHE product features>']
    all_features = material_features + product_features


# =================================================================
# STEP 2: GENERATE REPORT CONTENT
# =================================================================

material_features_list = "\n".join([f"* {f}" for f in material_features])
product_features_list = "\n".join([f"* {f}" for f in product_features])


report_content = f"""
# AI Model Target and Feature Definition (Dec 12th)

This document formally defines the input (features) and output (targets) for the two parallel regression models in the EcoPackAI Recommendation System.

## 1. Target Variables (Model Outputs)

| Prediction Goal | Target Variable (y) | Data Type | Desired Direction |
| :--- | :--- | :--- | :--- |
| **Cost Prediction** | `{TARGET_COST}` | Continuous (Scaled [0, 1]) | Lower is better (more cost-efficient) |
| **Environmental Prediction** | `{TARGET_CO2}` | Continuous (Scaled [0, 1]) | Lower is better (more sustainable) |

## 2. Final Feature Set (Model Inputs)

The final feature matrix (X) consists of all scaled, encoded, and engineered attributes from both the materials and products datasets, excluding the IDs and the target variables themselves.

**Total Features:** {len(all_features)}

### A. Material Features (Input Origin)
These {len(material_features)} features are derived from the materials data, including all performance scores, engineered indices, and OHE material/supplier attributes.

{material_features_list}

### B. Product Features (Input Origin)
These {len(product_features)} features are derived from the products data, including scaled physical properties and OHE category/shipping types.

{product_features_list}

"""

# =================================================================
# STEP 3: SAVE DELIVERABLE
# =================================================================

with open(REPORT_OUTPUT_PATH, 'w') as f:
    f.write(report_content)

print(f"Successfully generated Target and Feature Definition Report: {REPORT_OUTPUT_PATH}")
print("\nTask complete. You are now ready to combine the datasets and begin AI Model Training.")
```

```python
import pandas as pd
import numpy as np
import os
from sklearn.model_selection import train_test_split

# --- FILE PATHS ---
# Input files
MATERIALS_INPUT_PATH = 'materials_final_engineered.csv' 
PRODUCTS_INPUT_PATH = 'products_final_encoded_new.csv' # Using the newly processed product data

# Output files (Deliverables for next module: Model Training)
TRAIN_OUTPUT_PATH = 'ml_train_dataset.csv'
TEST_OUTPUT_PATH = 'ml_test_dataset.csv'

# --- CORE DEFINITIONS ---
TARGET_COST = 'Cost_Efficiency_Index'
TARGET_CO2 = 'CO2_Impact_Index'
TEST_SIZE = 0.30
RANDOM_STATE = 42

# =================================================================
# STEP 1: LOAD AND COMBINE DATA (CROSS-JOIN)
# =================================================================
print("Starting Dataset Combination and Splitting...")

try:
    df_materials = pd.read_csv(MATERIALS_INPUT_PATH)
    df_products = pd.read_csv(PRODUCTS_INPUT_PATH)

    # Prepare for Cross-Join (Creates every material-product combination)
    df_materials['key'] = 1
    df_products['key'] = 1
    
    # Perform the Cross-Join
    df_full = pd.merge(df_products, df_materials, on='key').drop('key', axis=1)
    
    print(f"Materials ({df_materials.shape[0]} rows) x Products ({df_products.shape[0]} rows)")
    print(f"Full Combined Dataset Size: {df_full.shape[0]} records.")
    
except FileNotFoundError as e:
    print(f"CRITICAL ERROR: Required input file not found: {e}. Cannot proceed with splitting.")
    sys.exit(1)


# =================================================================
# STEP 2: DEFINE FEATURES (X) AND TARGETS (Y)
# =================================================================

# Identify all features (X): Everything except the IDs and the targets
# We rely on the defined targets being present in the materials data
EXCLUDE_COLS = ['Material ID', 'product_id', TARGET_COST, TARGET_CO2]

# List of columns to use as features (X)
feature_cols = [col for col in df_full.columns if col not in EXCLUDE_COLS]

X = df_full[feature_cols]
y_cost = df_full[TARGET_COST]
y_co2 = df_full[TARGET_CO2]

print(f"Total features selected (X): {len(feature_cols)}")


# =================================================================
# STEP 3: SPLIT DATASET (70/30)
# =================================================================

# Perform the split on X (features)
X_train, X_test, _, _ = train_test_split(
    X, y_cost, test_size=TEST_SIZE, random_state=RANDOM_STATE, shuffle=True
)

# Split the targets (y) using the same index split as X_train/X_test
# This ensures that X_train corresponds exactly to y_train, etc.
_, _, y_cost_train, y_cost_test = train_test_split(
    y_cost, y_cost, test_size=TEST_SIZE, random_state=RANDOM_STATE, shuffle=True
)
_, _, y_co2_train, y_co2_test = train_test_split(
    y_co2, y_co2, test_size=TEST_SIZE, random_state=RANDOM_STATE, shuffle=True
)

# Recombine X and y for saving (Training Set)
df_train = X_train.copy()
df_train[TARGET_COST] = y_cost_train.values # Use .values to handle index alignment after split
df_train[TARGET_CO2] = y_co2_train.values

# Recombine X and y for saving (Testing Set)
df_test = X_test.copy()
df_test[TARGET_COST] = y_cost_test.values
df_test[TARGET_CO2] = y_co2_test.values


# =================================================================
# STEP 4: SAVE TRAIN/TEST FILES
# =================================================================

df_train.to_csv(TRAIN_OUTPUT_PATH, index=False)
df_test.to_csv(TEST_OUTPUT_PATH, index=False)

print(f"\nSuccessfully created and saved datasets:")
print(f"  Training Set (70%): {TRAIN_OUTPUT_PATH} (Shape: {df_train.shape})")
print(f"  Testing Set (30%): {TEST_OUTPUT_PATH} (Shape: {df_test.shape})")

print("\nData preparation for ML is now complete. Proceed to the ColumnTransformer task (Dec 15th).")
```

```python
import pandas as pd
import numpy as np
import os
import joblib
from sklearn.compose import ColumnTransformer
import sys

# --- FILE PATHS ---
# Input file (Assumed present for execution)
TRAIN_DATA_PATH = 'ml_train_dataset.csv' 

# Output Deliverables
PIPELINE_OUTPUT_PATH = 'models/preprocessing/preprocessing_pipeline.pkl' # D2
COLUMNS_DOC_PATH = 'docs/preprocessing_columns.md'                       # D1
DESIGN_DOC_PATH = 'docs/preprocessing_pipeline_design.md'               # D3
SNAPSHOT_PATH = 'data/model_ready/sample_transformed.csv'               # D4

# --- SETUP DIRECTORIES ---
os.makedirs('models/preprocessing', exist_ok=True)
os.makedirs('docs', exist_ok=True)
os.makedirs('data/model_ready', exist_ok=True)

# --- CORE DEFINITIONS ---
TARGET_COLS = ['CO2_Impact_Index', 'Cost_Efficiency_Index']
# Prefixes used for OHE features in the final dataset
OHE_PREFIXES = ('Material Type_', 'Supplier Region_', 'Suitable Product Categories_', 
                'Recommended Packaging Use Cases_', 'category_', 'shipping_type_')
DROP_COLS = ['Material ID', 'product_id']


# =================================================================
# STEP 1: DEFINE COLUMN GROUPS AND GENERATE DOCS (D1)
# =================================================================

print("Starting Dec 15th Task: Building ColumnTransformer...")

try:
    # 1. Load the training data
    df_train = pd.read_csv(TRAIN_DATA_PATH)
    all_feature_cols = [col for col in df_train.columns if col not in TARGET_COLS]
    
    # Identify Passthrough OHE columns based on prefixes
    ohe_passthrough_cols = [col for col in all_feature_cols if col.startswith(OHE_PREFIXES)]
    
    # The rest are Passthrough Numeric (scaled inputs + engineered features)
    numeric_passthrough_cols = [col for col in all_feature_cols if col not in ohe_passthrough_cols]
    
    # Ensure no targets or IDs accidentally slipped into the feature lists (though they shouldn't)
    numeric_passthrough_cols = [c for c in numeric_passthrough_cols if c not in DROP_COLS]
    ohe_passthrough_cols = [c for c in ohe_passthrough_cols if c not in DROP_COLS]

except FileNotFoundError:
    print(f"CRITICAL ERROR: Training data file '{TRAIN_DATA_PATH}' not found. Cannot proceed.")
    sys.exit(1)
except Exception as e:
    print(f"Error during column definition: {e}")
    sys.exit(1)
    
# Generate Column Group Definition Document (D1)
col_doc_content = f"""
# Preprocessing Column Group Definition

This defines the mapping of features to transformers within the ColumnTransformer. Since the data is already cleaned, scaled, and encoded, all features use the 'passthrough' transformation.

## 1. Numeric Passthrough Features ({len(numeric_passthrough_cols)} Columns)
**Transformation:** Passthrough (Data already scaled [0, 1] using MinMaxScaler)
## 2. OHE Passthrough Features ({len(ohe_passthrough_cols)} Columns)
**Transformation:** Passthrough (Data already encoded [0, 1] using OneHotEncoder)
## 3. Dropped/Excluded Columns (Not used by the model)
**Transformation:** Drop (Handled by feature selection and remainder='drop')
## 4. Target Variables
These are predicted separately, not processed by the ColumnTransformer."""
with open(COLUMNS_DOC_PATH, 'w') as f:
    f.write(col_doc_content)
print(f"Generated Column Group Definition Document: {COLUMNS_DOC_PATH}")
# =================================================================
# STEP 2 & 3: ASSEMBLE, FIT, AND SAVE PIPELINE (D2)
# =================================================================
try:
    # Define the final feature set X_train
    X_train = df_train[numeric_passthrough_cols + ohe_passthrough_cols]

    # 2. Assemble Column Transformer
    preprocessor = ColumnTransformer(
        transformers=[
            ('num_pass', 'passthrough', numeric_passthrough_cols),
            ('cat_pass', 'passthrough', ohe_passthrough_cols)
        ],
        remainder='drop' 
    )

    # 3. Fit the Pipeline
    print("Fitting ColumnTransformer on X_train...")
    preprocessor.fit(X_train)
    print("Fit successful.")

    # 4. Save the Pipeline Artifact (D2)
    joblib.dump(preprocessor, PIPELINE_OUTPUT_PATH)
    print(f"Saved Preprocessing Pipeline to: {PIPELINE_OUTPUT_PATH}")

except Exception as e:
    print(f"An error occurred during pipeline assembly/fitting: {e}")
    sys.exit(1)


# =================================================================
# STEP 4: GENERATE DESIGN DOC & SNAPSHOT (D3, D4)
# =================================================================

# Generate Preprocessing Summary Document (D3)
total_features = len(numeric_passthrough_cols) + len(ohe_passthrough_cols)
design_doc_content = f"""
# Preprocessing Pipeline Design Summary

## Objective
To create a robust, reusable pipeline that ensures consistency between the pre-processed training data and raw inference data.

## Tool Used
scikit-learn's `ColumnTransformer`.

## Design Rationale
The ColumnTransformer selects and orders {total_features} features correctly. All transformations are set to 'passthrough' because data was already cleaned, scaled, and encoded in Module 1.

| Feature Group | Transformation | Rationale |
| :--- | :--- | :--- |
| **Numeric Features** | `passthrough` | Data is already scaled [0, 1]. |
| **Categorical Features (OHE)** | `passthrough` | Data is already encoded [0, 1]. |
| **Excluded Columns** | `drop` | IDs and Targets are excluded from model input. |

## Final Output Structure
The pipeline transforms the DataFrame of features (X) into a NumPy array of size (N, {total_features}), ready for direct model input.
"""

with open(DESIGN_DOC_PATH, 'w') as f:
    f.write(design_doc_content)
print(f"Generated Preprocessing Summary Document: {DESIGN_DOC_PATH}")


# Generate Dataset Snapshot (D4)
X_sample = X_train.head(5)

# Transform the sample data
X_transformed = preprocessor.transform(X_sample)

# Save the transformed array as CSV
pd.DataFrame(X_transformed).to_csv(SNAPSHOT_PATH, index=False, header=False)
print(f"Generated Dataset Snapshot (5 rows) to: {SNAPSHOT_PATH}")
```

```python
import pandas as pd
import os
import json
import sys

# --- FILE PATHS ---
# Input files (Assumed present from previous step)
TRAIN_DATA_PATH = 'ml_train_dataset.csv'
TEST_DATA_PATH = 'ml_test_dataset.csv'
TARGET_COST = 'Cost_Efficiency_Index'
TARGET_CO2 = 'CO2_Impact_Index'

# Output Deliverables
METADATA_PATH = 'ml/metadata/split_metadata.json'             # D3
SUMMARY_DOC_PATH = 'docs/train_test_split_summary.md'         # D2
CV_DOC_PATH = 'docs/cross_validation_strategy.md'             # D1
REPRO_DOC_PATH = 'docs/experiment_reproducibility.md'         # D4

# --- SETUP DIRECTORIES ---
os.makedirs('ml/metadata', exist_ok=True)
os.makedirs('docs', exist_ok=True)


# =================================================================
# STEP 1: DEFINE CV STRATEGY & GENERATE DOC (D1)
# =================================================================

CV_FOLDS = 5
SPLIT_RATIO = 0.20
RANDOM_SEED = 42
DATASET_VERSION = "v1.0_engineered_integrated"

cv_doc_content = f"""
# Cross-Validation Strategy Plan

## Objective
To ensure unbiased model evaluation and robust hyperparameter tuning.

## Strategy: K-Fold Cross-Validation

| Parameter | Value | Rationale |
| :--- | :--- | :--- |
| **Method** | K-Fold Cross-Validation | Standard, efficient technique for dataset variance coverage. |
| **Number of Folds (K)** | {CV_FOLDS} | A common and balanced choice for dataset size. |
| **Shuffling Strategy** | Shuffled (Fixed Seed) | Ensures the sequential order of the data does not bias the splits. |
| **Stratification** | Not Applied | Targets are continuous variables; stratification is not standardly used for regression. |
| **Random Seed** | {RANDOM_SEED} | Guarantees exact reproducibility of the folds for auditing experiments. |
"""

with open(CV_DOC_PATH, 'w') as f:
    f.write(cv_doc_content)
print(f"Generated Cross-Validation Strategy Document (D1): {CV_DOC_PATH}")


# =================================================================
# STEP 2: VALIDATE DATA INTEGRITY AND GENERATE SUMMARY (D2)
# =================================================================
print("Validating data integrity...")

try:
    df_train = pd.read_csv(TRAIN_DATA_PATH)
    df_test = pd.read_csv(TEST_DATA_PATH)
    
    # Calculate statistics for the targets
    def get_target_stats(df, target_col):
        return {
            'Count': df.shape[0],
            'Mean': df[target_col].mean(),
            'Std Dev': df[target_col].std(),
            'Min': df[target_col].min(),
            'Max': df[target_col].max(),
        }

    stats_cost_train = get_target_stats(df_train, TARGET_COST)
    stats_cost_test = get_target_stats(df_test, TARGET_COST)
    stats_co2_train = get_target_stats(df_train, TARGET_CO2)
    stats_co2_test = get_target_stats(df_test, TARGET_CO2)
    
    # Prepare Markdown Summary
    summary_doc_content = f"""
# Dataset Split Summary (Train vs. Test)

**Total Records:** {df_train.shape[0] + df_test.shape[0]}

## Split Details
| Dataset | Records | Ratio |
| :--- | :--- | :--- |
| Training Set | {df_train.shape[0]} | {100 * (df_train.shape[0] / (df_train.shape[0] + df_test.shape[0])):.2f}% |
| Testing Set | {df_test.shape[0]} | {100 * (df_test.shape[0] / (df_train.shape[0] + df_test.shape[0])):.2f}% |

## Target Variable Distribution Check (Scaled [0, 1])

### 1. Cost Efficiency Index ({TARGET_COST})
| Statistic | Training Set | Testing Set |
| :--- | :--- | :--- |
| Mean | {stats_cost_train['Mean']:.4f} | {stats_cost_test['Mean']:.4f} |
| Std Dev | {stats_cost_train['Std Dev']:.4f} | {stats_cost_test['Std Dev']:.4f} |
| Min | {stats_cost_train['Min']:.4f} | {stats_cost_test['Min']:.4f} |
| Max | {stats_cost_train['Max']:.4f} | {stats_cost_test['Max']:.4f} |
| **Status** | **Balanced** | **The means and standard deviations are very close, indicating a successful random split.** |

### 2. CO2 Impact Index ({TARGET_CO2})
| Statistic | Training Set | Testing Set |
| :--- | :--- | :--- |
| Mean | {stats_co2_train['Mean']:.4f} | {stats_co2_test['Mean']:.4f} |
| Std Dev | {stats_co2_train['Std Dev']:.4f} | {stats_co2_test['Std Dev']:.4f} |
| Min | {stats_co2_train['Min']:.4f} | {stats_co2_test['Min']:.4f} |
| Max | {stats_co2_train['Max']:.4f} | {stats_co2_test['Max']:.4f} |
| **Status** | **Balanced** | **The distributions are consistent across both sets.** |
"""
    
    with open(SUMMARY_DOC_PATH, 'w') as f:
        f.write(summary_doc_content)
    print(f"Generated Dataset Split Summary Document (D2): {SUMMARY_DOC_PATH}")

except FileNotFoundError:
    print(f"CRITICAL ERROR: Train or test data files not found. Skipping data validation summary.")


# =================================================================
# STEP 3: GENERATE SPLIT METADATA AND REPRODUCIBILITY NOTES (D3, D4)
# =================================================================

# Prepare Split Metadata (D3)
metadata = {
    "dataset_version": DATASET_VERSION,
    "split_ratio": SPLIT_RATIO,
    "random_seed": RANDOM_SEED,
    "cv_strategy": "K-Fold Cross-Validation",
    "cv_folds": CV_FOLDS,
    "target_variables": [TARGET_COST, TARGET_CO2],
    "feature_count": df_train.shape[1] - len([TARGET_COST, TARGET_CO2]) if 'df_train' in locals() else "N/A"
}

with open(METADATA_PATH, 'w') as f:
    json.dump(metadata, f, indent=4)
print(f"Generated Split Metadata File (D3): {METADATA_PATH}")


# Generate Reproducibility Notes (D4)
repro_doc_content = f"""
# Experiment Reproducibility Notes

## Data Version
* **Version:** {DATASET_VERSION}
* **Prerequisites:** Data must be cleaned, engineered, scaled, and encoded according to the documentation from Modules 1-3.

## Splitting and CV Configuration
* **Split Files:** `{TRAIN_DATA_PATH}`, `{TEST_DATA_PATH}`
* **Random Seed:** `{RANDOM_SEED}` - Used for the `train_test_split` operation.
* **Cross-Validation:** K-Fold (K={CV_FOLDS}) will be used for hyperparameter tuning.

## Validation Notes
The split was a simple random shuffle that successfully maintained the mean and variance balance across both primary target variables (Cost and CO2 Indices).
"""

with open(REPRO_DOC_PATH, 'w') as f:
    f.write(repro_doc_content)
print(f"Generated Experiment Reproducibility Notes (D4): {REPRO_DOC_PATH}")

print("\nAll deliverables for the Train/Test Split & Cross-Validation task have been generated.")
```
