```python
# cell 4 Data Cleaning & Preprocessing

# 1. Check missing values (documentation purpose)
material.isnull().sum()
material.dropna(inplace=True)

# 2. Standardize column names (snake_case)
material.columns = material.columns.str.lower()

# 3. Ensure no duplicates after drop
material = material.drop_duplicates()

# print("\nAfter cleaning:")
material.shape
```

```python
product.isnull().sum()
product.columns = product.columns.str.lower()
product = product.drop_duplicates()
product.shape
```

```python
material.columns = (
    material.columns
    .str.strip()
    .str.lower()
    .str.replace(r"[^\w]+", "_",regex=True)
)
product.columns = (
    product.columns
    .str.strip()
    .str.lower()
    .str.replace(r"[^\w]+", "_",regex=True)
)
```

```python
material['cost_per_kg'] = material['cost_per_unit_usd_']/((material['total_material_weight_tons_']*1000)/material['annual_usage_units_'])
```

```python
material['biodegradation_percent'] = (365/material['biodegradation_time_days_'])
```

```python
material['weight_capacity'] = (material['strength_mpa'] * 0.5 + material['moisture_resistance_score'] * 0.25 + material['thermal_resistance_score'] * 0.25)
```

```python
material.columns
```

```python
material.drop(columns=['packaging_type',
       'recommended_packaging_use_cases', 'supplier_region',
       'recyclability_category', 'recycled_content_', 'reusability_',
       'biodegradation_time_days_', 'end_of_life_disposal_',
       'carbon_footprint_kg_co2_unit_',
       'waste_reduction_impact_', 'sustainability_target_progress_',
        'moisture_resistance_score', 'thermal_resistance_score',
       'cost_per_unit_usd_', 'annual_usage_units_',
       'total_material_weight_tons_', 'supplier_sustainability_compliance_'],
       inplace=True)
```

```python
material.rename(columns={'biodegradation_percent':'biodegradability_percent'},inplace=True)
material.rename(columns={'co2_emission_per_kg_estimated_':'co2_emission_score'},inplace=True)
material.rename(columns={'recyclability_':'recyclability_percent'},inplace=True)

```
