```python
# cell 2 load dataset

material = pd.read_csv(r'C:\Users\Sneha\OneDrive\Desktop\EcoPackAI\ml\data\raw\material_dataset.csv')
print(" material dataset \n")
print(material.head())
print(" \n product dataset \n")
product = pd.read_csv(r'C:\Users\Sneha\OneDrive\Desktop\EcoPackAI\ml\data\raw\Product.csv')
print(product.head())
```

```python
# cell 3 dataset info
print("\n Material Dataset Info \n")
material.info()
print("\n Product Dataset Info \n")
product.info()
```

```python
# cell 4 Data Cleaning & Preprocessing

# 1. Check missing values (documentation purpose)
print("Material missing values:\n", material.isnull().sum())
print("\nProduct missing values:\n", product.isnull().sum())

# 2. Standardize column names (snake_case)
material.columns = material.columns.str.lower()
product.columns = product.columns.str.lower()

# 3. Ensure no duplicates after drop
material = material.drop_duplicates()
product = product.drop_duplicates()

print("\nAfter cleaning:")
print(material.shape, product.shape)
```

```python
# cell 5 Save cleaned datasets
material.to_csv("../processed/material_cleaned.csv", index=False)
product.to_csv("../processed/product_cleaned.csv", index=False)
```
