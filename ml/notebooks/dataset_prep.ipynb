```python
# cell 4 Data Cleaning & Preprocessing

# 1. Check missing values (documentation purpose)
material.isnull().sum()
material.dropna(inplace=True)

# 2. Standardize column names (snake_case)
material.columns = material.columns.str.lower()

# 3. Ensure no duplicates after drop
material = material.drop_duplicates()

# print("\nAfter cleaning:")
material.shape
```

```python
product.isnull().sum()
product.columns = product.columns.str.lower()
product = product.drop_duplicates()
product.shape
```

```python
# cell 5 Save cleaned datasets
material.to_csv("material_cleaned.csv", index=False)
product.to_csv("product_cleaned.csv", index=False)
```
