Objective:
Define target variables and feature sets from validated datasets
to prepare model-ready inputs for downstream training.


```python
#cell 1 Environment setup 
import pandas as pd
```

```python
#cell 2 load dataset
material = pd.read_csv("../data/processed/material_engineered.csv")

print(material.shape)
material.head()
```

```python
#cell 3 Decide TARGET VARIABLE (y)
material["recommended_material_type"] = material["material_type"]
```

```python
# cell 4  select features (X)
feature_columns = [
    "strength_mpa",
    "weight_capacity",
    "biodegradability_percent",
    "co2_emission_score",
    "recyclability_percent",
    "cost_per_kg",
    "co2_impact_index",
    "cost_efficiency_index",
    "material_suitability_score"
]

X = material[feature_columns]
y = material["recommended_material_type"]

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
#cell 5 Final check
assert X.isnull().sum().sum() == 0
assert y.isnull().sum() == 0

print("Target & feature sanity checks passed")
```

```python
# cell 6 Saving features and target variable
X.to_csv("../data/final/X_raw.csv", index=False)
y.to_csv("../data/final/y_raw.csv", index=False)

print("X_raw.csv and y_raw.csv saved successfully")
```

```python
# cell 7 summary
summary = {
    "num_features": X.shape[1],
    "num_samples": X.shape[0],
    "target": "recommended_material_type",
    "features": feature_columns
}

summary
```
