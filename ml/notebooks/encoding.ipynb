```python
import pandas as pd

df = pd.read_csv(r'C:\Users\Ranjit\OneDrive\Desktop\AI-Powered-Sustainable-Packaging-Recommendation-System\data\processed\material_cleaned.csv')

categorical_cols = df.select_dtypes(include=["object"]).columns.tolist()
numerical_cols = df.select_dtypes(include=["int64", "float64"]).columns.tolist()

print("Categorical Columns:")
print(categorical_cols)

print("\nNumerical Columns:")
print(numerical_cols)
```

```python
from sklearn.preprocessing import OneHotEncoder


# Identify categorical columns
categorical_cols = df.select_dtypes(include=["object"]).columns.tolist()

print("Categorical columns to encode:")
print(categorical_cols)

# One-Hot Encoder
ohe = OneHotEncoder(handle_unknown="ignore", sparse_output=False)

# Fit & transform
encoded_array = ohe.fit_transform(df[categorical_cols])

# Create encoded DataFrame
encoded_df = pd.DataFrame(
    encoded_array,
    columns=ohe.get_feature_names_out(categorical_cols)
)

# Combine with numerical columns
numerical_df = df.drop(columns=categorical_cols).reset_index(drop=True)
final_encoded_df = pd.concat([numerical_df, encoded_df], axis=1)

print("Encoding completed. Final shape:", final_encoded_df.shape)
```

```python
from sklearn.preprocessing import MinMaxScaler

# Identify numeric columns (after encoding)
numeric_cols = final_encoded_df.select_dtypes(include=["int64", "float64"]).columns.tolist()

# Initialize scaler
scaler = MinMaxScaler()

# Fit & transform numeric columns
final_encoded_df[numeric_cols] = scaler.fit_transform(final_encoded_df[numeric_cols])

print("Normalization completed.")
print("Sample of normalized data:")
display(final_encoded_df[numeric_cols].head())
```

```python
import os
import joblib

# Create directories if not exist
os.makedirs("models/encoders", exist_ok=True)
os.makedirs("models/scalers", exist_ok=True)

# Save One-Hot Encoder
joblib.dump(ohe, "models/encoders/ohe_encoder.pkl")

# Save MinMax Scaler
joblib.dump(scaler, "models/scalers/numeric_scaler.pkl")

print("Encoders and scaler saved successfully.")
```

```python
import os

# Create output directory
os.makedirs("data/model_ready", exist_ok=True)

# Save final encoded dataset
final_encoded_df.to_csv(
    "materials_final_encoded.csv",
    index=False
)

print("Final ML-ready dataset exported successfully")
print("File: data/final/materials_final_encoded.csv")
```
