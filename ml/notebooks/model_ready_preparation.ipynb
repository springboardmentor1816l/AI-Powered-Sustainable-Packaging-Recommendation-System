```python
# cell 1 Environment setup
import pandas as pd
from sklearn.preprocessing import OneHotEncoder, MinMaxScaler
```

```python
#cell 2 Load Dataset
df = pd.read_csv("../data/processed/material_engineered.csv")
df.head()
```

```python
# cell 3 Separate Categorical & Numerical Columns
categorical_cols = ["material_type", "industry_use_case"]

numeric_cols = [
    "strength_mpa",
    "weight_capacity",
    "biodegradability_percent",
    "co2_emission_score",
    "recyclability_percent",
    "cost_per_kg",
    "co2_impact_index",
    "cost_efficiency_index",
    "material_suitability_score"
]
```

```python
# cell 4 One-Hot Encoding for Categorical Variables
encoder = OneHotEncoder(sparse_output=False, handle_unknown="ignore")

encoded_cat = encoder.fit_transform(df[categorical_cols])
encoded_cat_df = pd.DataFrame(
    encoded_cat,
    columns=encoder.get_feature_names_out(categorical_cols)
)
```

```python
#cell 5 Scale Numerical Features
scaler = MinMaxScaler()

scaled_num = scaler.fit_transform(df[numeric_cols])
scaled_num_df = pd.DataFrame(scaled_num, columns=numeric_cols)
```

```python
# cell 6 Combine FINAL MODEL-READY DATASET
model_ready_df = pd.concat(
    [scaled_num_df, encoded_cat_df],
    axis=1
)

model_ready_df.head()
```

```python
# cell 7 Save Model-Ready as Parquet Files

# Cleaned (before feature engineering)
pd.read_csv("../data/processed/material_cleaned.csv") \
  .to_parquet("../data/final/materials_cleaned.parquet", index=False)

# Engineered
df.to_parquet("../data/final/materials_engineered.parquet", index=False)

# Model-ready
model_ready_df.to_parquet("../data/final/materials_model_ready.parquet", index=False)

print("All Parquet files saved successfully")

```
